```python
idade = int(input("Digite sua idade: "))
if idade <= 12:
  print("Filme de classificação infatil")
elif idade < 18:
  print("Filme de classificação para adolescente")
elif idade < 60:
  print("Filme de classificação para adulto")
else:
  print("idoso")
```

```python
# Limite máximo de ingressos disponíveis no sistema
LIMITE_INGRESSOS = 10

# Inicializa os contadores para cada faixa etária
total_infantil = 0
total_adolescente = 0
total_adulto = 0
total_idoso = 0

print("--- Sistema de Venda de Ingressos com Validação de Documentos ---")
print(f"Capacidade máxima do evento: {LIMITE_INGRESSOS} ingressos.")
print("Digite a idade para comprar ou 'S' para encerrar e ver o relatório.")

while True:
    # Calcula o total vendido até o momento para gerenciar o equilíbrio
    total_atual = total_infantil + total_adolescente + total_adulto + total_idoso
    ingressos_restantes = LIMITE_INGRESSOS - total_atual

    # Se atingir o limite de 10 ingressos, o programa encerra automaticamente
    if ingressos_restantes == 0:
        print("\n⚠️ LIMITE DE 10 INGRESSOS ATINGIDO! Encerrando o sistema...")
        break

    print(f"\n[Ingressos restantes no estoque: {ingressos_restantes}]")
    entrada = input("Digite sua idade (ou 'S' para sair): ").strip()

    # Condição para saída manual
    if entrada.upper() == 'S':
        break

    try:
        idade = int(entrada)

        if idade < 0:
            print("Idade inválida! Digite um valor maior ou igual a 0.")
            continue

        # Determina a categoria e o documento necessário antes de computar o ingresso
        if idade <= 12:
            categoria = "infantil"
            documento_obrigatorio = "RG ou Certidão de Nascimento (acompanhado dos pais)"
        elif idade < 18:
            categoria = "adolescente"
            documento_obrigatorio = "RG ou Documento de Identidade Estudantil"
        elif idade < 60:
            categoria = "adulto"
            documento_obrigatorio = "RG ou CNH"
        else:
            categoria = "idoso"
            documento_obrigatorio = "RG (comprovação de idade para prioridade/estatuto)"

        # Solicita a verificação do documento necessário
        print(f"\n[Validação] Categoria: {categoria.capitalize()}")
        print(f"Documento obrigatório necessário: {documento_obrigatorio}")

        # Loop para forçar apenas resposta 'S' ou 'N'
        while True:
            confirmacao = input("O cliente apresentou o documento válido? (S para Sim / N para Não): ").strip().upper()
            if confirmacao in ['S', 'N']:
                break
            print("Resposta inválida! Digite apenas 'S' ou 'N'.")

        if confirmacao == 'S':
            # Se o documento for válido, computa o ingresso na categoria correspondente
            if categoria == "infantil":
                print("Filme de classificação infantil. Ingresso emitido!")
                total_infantil += 1
            elif categoria == "adolescente":
                print("Filme de classificação para adolescente. Ingresso emitido!")
                total_adolescente += 1
            elif categoria == "adulto":
                print("Filme de classificação para adulto. Ingresso emitido!")
                total_adulto += 1
            elif categoria == "idoso":
                print("Idoso. Ingresso emitido!")
                total_idoso += 1
        else:
            print("❌ Venda cancelada! Entrada proibida sem a apresentação do documento obrigatório.")

    except ValueError:
        print("Entrada inválida! Por favor, digite apenas números inteiros ou 'S' para sair.")

# --- BLOCO REAVALIADO: Geração e salvamento do relatório final ---
total_geral = total_infantil + total_adolescente + total_adulto + total_idoso
porcentagem_ocupacao = (total_geral / LIMITE_INGRESSOS) * 100

# Construção limpa e estruturada do relatório (sem acúmulos ou repetições)
relatorio_linhas = [
    "=" * 40,
    "           RELATÓRIO DE VENDAS          ",
    "=" * 40,
    f"  Ingressos Infantis:     {total_infantil}",
    f"  Ingressos Adolescentes: {total_adolescente}",
    f"  Ingressos Adultos:      {total_adulto}",
    f"  Ingressos Idosos:       {total_idoso}",
    "-" * 40,
    f"  Total Vendido:          {total_geral} / {LIMITE_INGRESSOS}",
    f"  Taxa de Ocupação:       {porcentagem_ocupacao:.1f}%",
    "=" * 40
]

# Junta as linhas em uma única string com quebras de linha
relatorio_final = "\n".join(relatorio_linhas)

# Exibe o relatório formatado na tela
print("\n" + relatorio_final)

# Salva em arquivo sobrescrevendo qualquer versão anterior de forma limpa
try:
    with open("relatorio_vendas.txt", "w", encoding="utf-8") as arquivo:
        arquivo.write(relatorio_final)
    print("\n💾 Relatório salvo com sucesso em 'relatorio_vendas.txt'!")
except Exception as e:
    print(f"\n⚠️ Erro ao salvar o arquivo: {e}")
```

```python
import pandas as pd

def get_divisors(n):
    return [i for i in range(1, n + 1) if n % i == 0]

# Lista de números
numeros = range(1, 11)

# Criando os dados da tabela
dados = []

for num in numeros:
    divisores = get_divisors(num)

    dados.append({
        "Número": num,
        "Divisores": str(divisores),
        "É primo?": "✅ Sim" if len(divisores) == 2 else "❌ Não"
    })

# Criando a tabela
tabela = pd.DataFrame(dados)

# Exibindo a tabela
display(tabela)
```

```python
# Funções defunidas pelo user.
def calcular_desconto(valor, percentual):
    """calcula o valor final aplicando um desconto percentual."""
    if percentual <0 or percentual > 100:
      return None  # marcacao invalida
    desconto = valor * (percentual/100)
    return valor - desconto
def registrar_venda(produto, valor_final):
  """imprime o reguistro de uma venda."""
  print(f">> venda registrada: {produto} por R${valor_final:.2f}")

#funcao lambda para arredondar
arredondar = lambda valor: round(valor, 2)
```

```python
#variaveis principais do PDV
total_vendas = 0
qnt_vendas = 0
```

```python
#loop principal do caixa. LOJINHAPDV
while True:
  print("\n----- minha loja - pdv-----")
  print("1 - registro de vendas")
  print("2 - relatorio diario")
  print("3 - encerrar caixa")
  opcao = input("digite a opcao desejada: ")
  if opcao == "0":
    print("encerrando o caixa. ate amanha")
    break

  elif opcao == "1":
    produto = input("nome do produto")
    valor = float(input("valor do produto: R$ "))
    percentual = float(input("desconto (em %): "))
    valor_final = calcular_desconto(valor, percentual)
    if valor_final is None:
      print("desconto invalido use um valor entre 0 e 100.")
    else:
      valor_final = arredondar(valor_final)
      registrar_venda(produto, valor_final)
      total_vendas = total_vendas + valor_final
      qnt_vendas = qnt_vendas + 1
  elif opcao == "2":
    print(f"\ntotal de vendas do dia: R${qnt_vendas}")
    print(f"total faturado: R${arredondar(total_vendas)}")
  else:
    print("opcao invalida. tente novamente")
```
